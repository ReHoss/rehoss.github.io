## Part I — Expected total discounted reward and policy evaluation

Throughout the sheet, $(\mathcal{S},\mathcal{A},P,r)$ is a finite Markov decision process in the sense of Practical Session 1: the state space $\mathcal{S}$ and every set $\mathcal{A}(s)$ of admissible actions are finite, and $\mathcal{K}$ is the set of admissible state–action pairs. Decision rules are defined as in Practical Session 1, and a policy is now an infinite sequence $\pi=(d_0,d_1,\dots)$ of decision rules. The set of all policies, history-dependent and randomised ones included, is $\Pi$. The law of the process started at $s\in\mathcal{S}$ under $\pi$ is $\mathbb{P}^{\pi}_{s}$, and its expectation is $\mathbb{E}^{\pi}_{s}$. The discount factor $\gamma\in[0,1)$ is fixed, and $\lVert r\rVert_{\infty}=\max_{(s,a)\in\mathcal{K}}|r(s,a)|$.

**Definition 1 (Expected total discounted reward; Puterman 1994, Section 5.1, page 121).** 
The *expected total discounted reward* of a policy $\pi\in\Pi$ started at $s\in\mathcal{S}$ is

$$
V^{\pi}_{\gamma}(s)=\lim_{N\to\infty}\mathbb{E}^{\pi}_{s}\Big[\sum_{t=0}^{N-1}\gamma^{t}r(S_t,A_t)\Big].
$$

To restrict the search for good policies to policies that depend on the current state only, Theorem 1 constructs, for each policy and each initial state, a randomised Markovian policy under which the pair (state, action) has the same law at every epoch.

**Theorem 1 (Reduction to Markovian policies; Puterman 1994, Theorems 5.5.1 and 5.5.3, pages 134 and 137).** 
Let $\pi\in\Pi$ and $s\in\mathcal{S}$. There exists a randomised Markovian policy $\pi'$, which may depend on $s$, such that

$$
\mathbb{P}^{\pi'}_{s}(S_t=s',A_t=a)=\mathbb{P}^{\pi}_{s}(S_t=s',A_t=a)
$$

for every $t\in\mathbb{N}$ and every $(s',a)\in\mathcal{K}$. In particular, $V^{\pi'}_{\gamma}(s)=V^{\pi}_{\gamma}(s)$.

The policy $\pi'$ depends on the initial state $s$ (Puterman, page 135), and it may have to be randomised even when $\pi$ is deterministic (page 136). Theorem 1 therefore gives one Markovian policy for each initial state, and no single Markovian policy with the value of $\pi$ at every state. Its consequence is an equality of suprema, state by state: with $\Pi^{\mathrm{MR}}$ the set of randomised Markovian policies,

$$
\sup_{\pi\in\Pi}V^{\pi}_{\gamma}(s)=\sup_{\pi\in\Pi^{\mathrm{MR}}}V^{\pi}_{\gamma}(s),\qquad s\in\mathcal{S}
$$

(Puterman, page 143).

To write the value of a Markovian policy in matrix form, let $d$ be a randomised Markovian decision rule, and write $d(a\mid s)$ for the probability that the distribution $d(s)$ gives to the action $a$. The reward vector and the transition matrix of $d$ are defined by

$$
r_{d}(s)=\sum_{a\in\mathcal{A}(s)}d(a\mid s)\,r(s,a),\qquad
 P_{d}(s'\mid s)=\sum_{a\in\mathcal{A}(s)}d(a\mid s)\,P(s'\mid s,a),\qquad s,s'\in\mathcal{S}
$$

(Puterman, equation (5.6.2), page 137). Let $\mathcal{D}$ be the set of deterministic Markovian decision rules, that is of the maps $d:\mathcal{S}\to\mathcal{A}$ with $d(s)\in\mathcal{A}(s)$ for every $s\in\mathcal{S}$. For $d\in\mathcal{D}$, $r_{d}(s)=r(s,d(s))$ and $P_{d}(s'\mid s)=P(s'\mid s,d(s))$.

To evaluate a Markovian policy $\pi=(d_0,d_1,\dots)$, the reward of the first epoch is separated from the rest. Expanding the first two epochs gives, the sums running over the admissible actions and over $\mathcal{S}$,

$$
V^{\pi}_{\gamma}(s)=\sum_{a}d_0(a\mid s)\Big[r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)\sum_{a'}d_1(a'\mid s')\Big(r(s',a')+\gamma\sum_{s''}P(s''\mid s',a')\cdots\Big)\Big].
$$

The factor of $\gamma\,P(s'\mid s,a)$ is the expected total discounted reward, from the state $s'$, of the shifted policy $\pi'=(d_1,d_2,\dots)$. Hence

$$
V^{\pi}_{\gamma}(s)=r_{d_0}(s)+\gamma\sum_{s'\in\mathcal{S}}P_{d_0}(s'\mid s)\,V^{\pi'}_{\gamma}(s'),\qquad s\in\mathcal{S}
\tag{1}
$$

(Puterman, equation (6.1.4), page 144). The discounted reward of $\pi$ is that of a one-epoch problem in which the decision maker uses $d_0$ and receives, as terminal reward, the discounted reward of $\pi'$ from the next state.

When $\pi=d^{\infty}=(d,d,\dots)$ is stationary, the shifted policy is $\pi$ itself, and (1) becomes an equation whose unknown is the function $V^{d^{\infty}}_{\gamma}$. For a function $v:\mathcal{S}\to\mathbb{R}$, define $\mathcal{T}^{d}v:\mathcal{S}\to\mathbb{R}$ by

$$
(\mathcal{T}^{d}v)(s)=r_{d}(s)+\gamma\sum_{s'\in\mathcal{S}}P_{d}(s'\mid s)\,v(s'),\qquad s\in\mathcal{S},
$$

that is $\mathcal{T}^{d}v=r_{d}+\gamma P_{d}v$, a function $v$ being identified with the vector $(v(s))_{s\in\mathcal{S}}$. The map $\mathcal{T}^{d}$ is the *policy evaluation operator* of $d$. For $\pi=d^{\infty}$, equation (1) states that $V^{d^{\infty}}_{\gamma}$ is a fixed point of $\mathcal{T}^{d}$.

To compute the value of a stationary policy, Theorem 2 solves this fixed-point equation, a linear system with one equation per state: the matrix $I-\gamma P_{d}$, where $I$ is the identity matrix indexed by $\mathcal{S}$, is invertible because $P_{d}$ is a stochastic matrix and $\gamma<1$.

**Theorem 2 (Policy evaluation; Puterman 1994, Theorem 6.1.1, page 145).** 
Let $d$ be a randomised Markovian decision rule. Then $V^{d^{\infty}}_{\gamma}$ is the unique function $v:\mathcal{S}\to\mathbb{R}$ such that $v=\mathcal{T}^{d}v$, and

$$
V^{d^{\infty}}_{\gamma}=(I-\gamma P_{d})^{-1}r_{d}.
$$

## Part II — Optimality equation

The question of Part II is the best expected total discounted reward and the policies that attain it. By Theorem 1, the supremum over all policies may be computed over randomised Markovian policies.

**Definition 2 (Value and optimal policy).** 
The *value* of the discounted problem is $V^{\ast}_{\gamma}(s)=\sup_{\pi\in\Pi}V^{\pi}_{\gamma}(s)$ for $s\in\mathcal{S}$. A policy $\pi^{\ast}\in\Pi$ is *optimal* when $V^{\pi^{\ast}}_{\gamma}(s)=V^{\ast}_{\gamma}(s)$ for every $s\in\mathcal{S}$.

For a function $v:\mathcal{S}\to\mathbb{R}$, define $\mathcal{T}^{\ast}v:\mathcal{S}\to\mathbb{R}$ by

$$
(\mathcal{T}^{\ast}v)(s)=\max_{a\in\mathcal{A}(s)}\Big\{r(s,a)+\gamma\sum_{s'\in\mathcal{S}}P(s'\mid s,a)\,v(s')\Big\},\qquad s\in\mathcal{S},
$$

the maximum existing because $\mathcal{A}(s)$ is finite. The map $\mathcal{T}^{\ast}$ is the *optimality operator*, and the equation $v=\mathcal{T}^{\ast}v$ is the *optimality equation*, or *Bellman equation*, of the discounted problem. It has the form of the finite-horizon optimality equations of Practical Session 1, with the factor $\gamma$ in front of the expected next value and the same unknown on both sides.

To solve the discounted problem through its optimality equation, Theorem 3 shows that this equation has exactly one solution, the value. Its proof shows that $\mathcal{T}^{\ast}$ is a contraction with modulus $\gamma$ for the norm $\lVert v\rVert_{\infty}=\max_{s\in\mathcal{S}}|v(s)|$, that is $\lVert \mathcal{T}^{\ast} u-\mathcal{T}^{\ast}v\rVert_{\infty}\le\gamma\lVert u-v\rVert_{\infty}$ for all $u,v:\mathcal{S}\to\mathbb{R}$ (Puterman, Proposition 6.2.4, page 151), and then applies the Banach fixed-point theorem (Puterman, Theorem 6.2.3, page 150).

**Theorem 3 (Optimality equation; Puterman 1994, Theorem 6.2.5(a), page 151).** 
The value $V^{\ast}_{\gamma}$ is the unique function $v:\mathcal{S}\to\mathbb{R}$ such that $v=\mathcal{T}^{\ast}v$.

For $(s,a)\in\mathcal{K}$, let

$$
Q^{\ast}_{\gamma}(s,a)=r(s,a)+\gamma\sum_{s'\in\mathcal{S}}P(s'\mid s,a)\,V^{\ast}_{\gamma}(s')
$$

be the *optimal action value* of $(s,a)$: the reward of the action $a$ in $s$ plus the discounted expected value of the next state. By Theorem 3, $V^{\ast}_{\gamma}(s)=\max_{a\in\mathcal{A}(s)}Q^{\ast}_{\gamma}(s,a)$ for every $s\in\mathcal{S}$.

To obtain an optimal policy from the value, Theorem 4 selects in each state an action that attains the maximum of $Q^{\ast}_{\gamma}$. Puterman calls such a decision rule *conserving* (page 152). For a conserving $d^{\ast}$, $\mathcal{T}^{d^{\ast}}V^{\ast}_{\gamma}=\mathcal{T}^{\ast}V^{\ast}_{\gamma}=V^{\ast}_{\gamma}$, so $V^{\ast}_{\gamma}$ is the value of $(d^{\ast})^{\infty}$ by Theorem 2.

**Theorem 4 (Optimal stationary deterministic policies; Puterman 1994, Theorem 6.2.7 and Corollary 6.2.8, pages 153–154).** 
Let $d^{\ast}\in\mathcal{D}$ satisfy

$$
d^{\ast}(s)\in\mathop{\mathrm{arg\,max}}\limits_{a\in\mathcal{A}(s)}Q^{\ast}_{\gamma}(s,a)\qquad\text{for every }s\in\mathcal{S}.
$$

Then the stationary policy $(d^{\ast})^{\infty}$ is optimal. Such a decision rule exists, so there exists an optimal policy that is stationary and deterministic.

Puterman states Theorem 4 under the condition that the maximum in the optimality equation is attained, which holds here because every $\mathcal{A}(s)$ is finite (Puterman, Theorem 6.2.10(a), page 154).

## Part III — Value iteration and policy iteration

By Theorem 3, solving the discounted problem amounts to computing the fixed point of the contraction $\mathcal{T}^{\ast}$. Value iteration applies $\mathcal{T}^{\ast}$ repeatedly. Policy iteration alternates the exact evaluation of a stationary policy (Theorem 2) and its improvement. Both are applied to the inventory problem of Practical Session 1, now over an infinite horizon.

**Example 2 (Inventory problem; Puterman 1994, Section 3.2).** 
The stock $s\in\mathcal{S}=\{0,1,\dots,M\}$ of a warehouse of capacity $M=3$ is observed at the beginning of each month, and $a\in\mathcal{A}(s)=\{0,1,\dots,M-s\}$ units are ordered and delivered at once. The demands of the months are independent, with $p_{0}=\tfrac14$, $p_{1}=\tfrac12$, $p_{2}=\tfrac14$, and $p_{j}=0$ for $j\ge3$. A demand that exceeds the stock is lost, so the next stock is $\left[s+a-D\right]^{+}$. The reward is $r(s,a)=F(s+a)-O(a)-h(s+a)$, with the expected revenue $F(0)=0$, $F(1)=6$, $F(2)=F(3)=8$, the ordering cost $O(0)=0$ and $O(a)=4+2a$ for $a\ge1$, and the holding cost $h(u)=u$. These are the data of Practical Session 1, without terminal value.

The convergence statement of value iteration uses a weaker notion than optimality. For $\varepsilon\in(0,+\infty)$, a policy $\pi$ is $\varepsilon$-*optimal* when $V^{\pi}_{\gamma}(s)\ge V^{\ast}_{\gamma}(s)-\varepsilon$ for every $s\in\mathcal{S}$ (Puterman, equation (6.2.17), page 156).

**Value iteration algorithm (Puterman 1994, Section 6.3.2, page 161), for $\gamma\in(0,1)$.**

1. Choose $v_{0}:\mathcal{S}\to\mathbb{R}$ and a tolerance $\varepsilon\in(0,+\infty)$, and set $n=0$.
2. Compute $v_{n+1}=\mathcal{T}^{\ast}v_{n}$, that is, for every $s\in\mathcal{S}$,
  
$$
v_{n+1}(s)=\max_{a\in\mathcal{A}(s)}\Big\{r(s,a)+\gamma\sum_{s'\in\mathcal{S}}P(s'\mid s,a)\,v_{n}(s')\Big\}.
$$

3. If $\lVert v_{n+1}-v_{n}\rVert_{\infty}<\varepsilon(1-\gamma)/(2\gamma)$, go to step 4. Otherwise replace $n$ by $n+1$ and return to step 2.
4. For every $s\in\mathcal{S}$, choose $d_{\varepsilon}(s)\in\mathop{\mathrm{arg\,max}}\limits_{a\in\mathcal{A}(s)}\big\{r(s,a)+\gamma\sum_{s'\in\mathcal{S}}P(s'\mid s,a)\,v_{n+1}(s')\big\}$, and stop.

To control the policy returned at step 4, Theorem 5 combines the contraction property of $\mathcal{T}^{\ast}$ with the stopping rule of step 3: the last iterate lies within $\varepsilon/2$ of the value, and the greedy policy loses at most $\varepsilon$.

**Theorem 5 (Convergence of value iteration; Puterman 1994, Theorem 6.3.1, page 161).** 
Let $\gamma\in(0,1)$, $v_{0}:\mathcal{S}\to\mathbb{R}$ and $\varepsilon\in(0,+\infty)$, and let $v_{n+1}=\mathcal{T}^{\ast}v_{n}$ for $n\in\mathbb{N}$. Then

**(a)** $\lVert v_{n}-V^{\ast}_{\gamma}\rVert_{\infty}$ tends to $0$ as $n\to\infty$,

**(b)** there exists $n_{0}\in\mathbb{N}$ such that the stopping rule of step 3 holds for every $n\ge n_{0}$,

**(c)** the stationary policy $d_{\varepsilon}^{\infty}$ defined at step 4 is $\varepsilon$-optimal,

**(d)** $\lVert v_{n+1}-V^{\ast}_{\gamma}\rVert_{\infty}<\varepsilon/2$ whenever the stopping rule of step 3 holds at $n$.

Policy iteration replaces the repeated application of $\mathcal{T}^{\ast}$ by exact evaluations, each one followed by a greedy improvement.

**Policy iteration algorithm (Puterman 1994, Section 6.4.1, pages 174–175).**

1. Choose $d_0\in\mathcal{D}$ and set $n=0$.
2. (Policy evaluation.) Compute $V^{d_n^{\infty}}_{\gamma}$ by solving $(I-\gamma P_{d_n})v=r_{d_n}$.
3. (Policy improvement.) Choose $d_{n+1}\in\mathcal{D}$ such that, for every $s\in\mathcal{S}$,
  
$$
d_{n+1}(s)\in\mathop{\mathrm{arg\,max}}\limits_{a\in\mathcal{A}(s)}\Big\{r(s,a)+\gamma\sum_{s'\in\mathcal{S}}P(s'\mid s,a)\,V^{d_n^{\infty}}_{\gamma}(s')\Big\},
$$

  keeping $d_{n+1}(s)=d_n(s)$ whenever $d_n(s)$ attains the maximum.
4. If $d_{n+1}=d_n$, stop and return $d_n$. Otherwise replace $n$ by $n+1$ and return to step 2.

Puterman's step 3 ends with “setting $d_{n+1}=d_n$ if possible” (page 175), a rule “included to avoid cycling”. Keeping the action of $d_n$ in every state where it attains the maximum satisfies this rule. Without it, when two actions attain the maximum, the algorithm may alternate between two optimal decision rules, and the stopping test of step 4 is then never met.

To prove that policy iteration stops, Theorem 6 uses the monotonicity of the values of the successive policies, $V^{d_{n+1}^{\infty}}_{\gamma}\ge V^{d_n^{\infty}}_{\gamma}$ (Puterman, Proposition 6.4.1, page 175), and the finiteness of $\mathcal{D}$.

**Theorem 6 (Convergence of policy iteration; Puterman 1994, Theorem 6.4.2, page 176).** 
The policy iteration algorithm stops after finitely many iterations. At termination, $V^{d_n^{\infty}}_{\gamma}$ is a solution of the optimality equation, and $d_n^{\infty}$ is optimal.
